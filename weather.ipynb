```python
import requests

# ============================================
# OpenWeather API Key
# Replace with your own API key
# ============================================

API_KEY = "YOUR_API_KEY"

BASE_URL = "https://api.openweathermap.org/data/2.5/weather"


def get_weather(city):
    """
    Fetch weather data for a given city.

    Parameters:
        city (str): Name of the city

    Returns:
        dict: Weather information if successful
        None: If city not found or API request fails
    """

    url = f"{BASE_URL}?q={city}&appid={API_KEY}&units=metric"

    try:
        response = requests.get(url)

        if response.status_code != 200:
            return None

        data = response.json()

        weather = {
            "city": data["name"],
            "country": data["sys"]["country"],
            "temperature": data["main"]["temp"],
            "feels_like": data["main"]["feels_like"],
            "humidity": data["main"]["humidity"],
            "pressure": data["main"]["pressure"],
            "wind_speed": data["wind"]["speed"],
            "weather": data["weather"][0]["main"],
            "description": data["weather"][0]["description"].title(),
            "icon": data["weather"][0]["icon"],
            "visibility": data.get("visibility", 0) / 1000,
            "clouds": data["clouds"]["all"],
            "latitude": data["coord"]["lat"],
            "longitude": data["coord"]["lon"],
            "sunrise": data["sys"]["sunrise"],
            "sunset": data["sys"]["sunset"]
        }

        return weather

    except Exception as e:
        print("Error:", e)
        return None


# ============================================
# Test the Program
# ============================================

if __name__ == "__main__":

    city = input("Enter City Name: ")

    result = get_weather(city)

    if result:

        print("\n========== Weather Report ==========")
        print("City        :", result["city"])
        print("Country     :", result["country"])
        print("Temperature :", result["temperature"], "°C")
        print("Feels Like  :", result["feels_like"], "°C")
        print("Humidity    :", result["humidity"], "%")
        print("Pressure    :", result["pressure"], "hPa")
        print("Wind Speed  :", result["wind_speed"], "m/s")
        print("Weather     :", result["weather"])
        print("Description :", result["description"])
        print("Visibility  :", result["visibility"], "km")
        print("Clouds      :", result["clouds"], "%")
        print("Latitude    :", result["latitude"])
        print("Longitude   :", result["longitude"])
        print("====================================")

    else:
        print("City not found or API request failed.")
```
